```python
df = df.drop_duplicates()

print("dataset shape after removing duplicates.",df.shape)
```

```python
print(df.isnull().sum())
```

```python
print("total missing values",df.isnull().sum().sum())
```

```python
missing_data = df.isnull().sum()
missing_data = missing_data[missing_data>0]
print("columns with missing values.")
print(missing_data)
```

```python
missing_percentage = (df.isnull().sum()/len(df))*100
print(missing_percentage[missing_percentage>0].round(2))
```

```python
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print("Remaining missing values:", df.isnull().sum().sum())
print("Dataset Shape:", df.shape)
```

```python
print(df.columns.tolist())
```

```python
print("Average Delivery Time:",
      round(df['Time_taken (min)'].mean(), 2), "minutes")
```

```python
print("Minimum Delivery Time:",
      df['Time_taken (min)'].min(), "minutes")

print("Maximum Delivery Time:",
      df['Time_taken (min)'].max(), "minutes")
```

```python
print("Median Delivery Time:",
      df['Time_taken (min)'].median(), "minutes")
```

```python
print(df['Time_taken (min)'].describe())
```

```python
weather_analysis = df.groupby('Weather_conditions')['Time_taken (min)'].mean()

print(weather_analysis.round(2))
```

```python
print(df['Weather_conditions'].value_counts())
```

```python
print("Highest Delivery Time Weather:",
      weather_analysis.idxmax())

print("Average Delivery Time:",
      round(weather_analysis.max(), 2), "minutes")
```

```python
print("Lowest Delivery Time Weather:",
      weather_analysis.idxmin())

print("Average Delivery Time:",
      round(weather_analysis.min(), 2), "minutes")
```

```python
traffic_analysis = df.groupby('Road_traffic_density')['Time_taken (min)'].mean()

print(traffic_analysis.round(2))
```

```python
print(df['Road_traffic_density'].value_counts())
```

```python
print("Highest Traffic Category:", traffic_analysis.idxmax())

print("Average Delivery Time:",
      round(traffic_analysis.max(), 2), "minutes")
```

```python
print("Lowest Traffic Category:", traffic_analysis.idxmin())

print("Average Delivery Time:",
      round(traffic_analysis.min(), 2), "minutes")
```
